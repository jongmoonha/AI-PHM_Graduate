##### numpy.arange([start, ]stop, [step, ]dtype=None, *, like=None)

```python
b = np.arange(0.1, 1, 0.1)
print(b)
```

##### Matrix Size 알아보기

```python
print(np.shape(b))
print('Size of Matrix is', np.shape(b))
```

##### Mean, var, kurtosis

```python
mean = np.mean(b);print(mean)
var = np.var(b);print(var)
kurt = np.kurtosis(b);print(kurt)
```

# 위에서 정의한 b 의 Kurtosis 를 구해보자.

Let's try to calculate kurtosis of b

```python
import scipy.stats
kurt = scipy.stats.kurtosis(b);print(kurt)
```

##### Sine Plot

```python
import matplotlib.pyplot as plt
```

```python
f = 1
fs  = 10
T = 2
t=np.arange(0, T, 1/fs)
print(np.shape(t))
x=np.cos(2*np.pi*f*t)
```

t를 x축으로, x를 y축으로 그래프를 그리고 y 축 범위를 [-1.1 ~ 1.1] 로 설정

```python
plt.figure(figsize=(6,4))
plt.plot(t, x, 'b-o')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.ylim([-1.1, 1.1])
plt.title('Sine Wave')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(4,4))
plt.plot(t,x,'r-o')
plt.xlabel('Time(s)')
plt.ylim([-1.1, 1.1])
plt.show()
```
